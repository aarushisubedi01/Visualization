## **Bar Graph**

```python
# Bar Graph
# Relationship between Numerical and Categorical Variable
# Hatch is used in barplot which is called patterns

# Hatch is a way to fill a chart with a pattern instead of a simple color.
```

```python
year_data = df.loc[11][:1]
year_data

colors = ['#E3C6AD', '#C8A98E', '#A67C68', '#8D6E63', '#D4B8A3',
          '#BCA492', '#9C7B6B', '#E5D1C5', '#C9B2A5', '#7D5A50']

hatches = ['/', '\\', '|', '-', '+', 'x', 'o', 'O', '.', '*']

plt.figure(figsize=(20,10))

bars = plt.bar(df.columns[1:], year_data, color= colors)

for bar, hatch in zip(bars, hatches):
  bar.set_hatch(hatch)

plt.xlabel('Country')
plt.ylabel('Price of Petrol')
plt.title('Petrol Price in USA over the years', {'fontsize':15, 'color':'red'})
```

## BoxPlot

```python
plt.style.use('seaborn-v0_8-dark-palette')

import seaborn as sns

plt.figure(figsize=(12,6))

plt.boxplot([df['Canada'], df['Japan'], df['USA']],
            labels = ['Canada', 'Japan', 'USA'])
plt.title('Boxplot rates for different countries')
plt.ylabel('Rates')
plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.boxplot([df['Canada'], df['Japan'], df['USA']],
            labels = ['Canada', 'Japan', 'USA'],
            patch_artist = True, # It allows the boxes in the box plot to be filled with color rather than just outlined.
            boxprops=dict(facecolor ='lightblue', color='Blue'), # It defines the properties of the box
            # Facecolor specifies that the interior of the boxes will be light blue color.
            # color sets the outline of the boxes to the blue.
            medianprops=dict(color='red'),
            whiskerprops=dict(color='green'), # It customizes the whiskers of the box plot # Indicates quartiles
            capprops=dict(color='orange') # It customizes the caps of the end of whiskers. # Indicates maximum and minimum value excluding outliers
            )
plt.title('Boxplot rates for different countries')
plt.ylabel('Rates')
plt.grid(True)
plt.show()
```

### **Histogram**

```python
df.drop(columns='Year').plot(kind = 'hist', alpha=0.5, bins = 10, figsize=(14,8), edgecolor='black')
# This sets the transparency of the histogram bars to 70% so they are slightly see-through - 'alpha'
# Bins divides the data 10 bin(categories) for the histogram.
# This specifies that the edges of the bars in the histogram will be colored black, making them standout

plt.title('Histogram of Rates for All Countries(1990-2008)')
plt.xlabel('Rate')
plt.ylabel('Frequency')
plt.legend(title ='Country', bbox_to_anchor=(1.09, 0.10), loc ='upper left')

# bbox_to_anchor = It further adjust the position of the legend, moving it slightly outside ther plot area horizontally but keeping it vertically alignes at about 90% height of uour histogram plot

#Legend Locations

# 1. upper right= place the legend in the upper right corner
# 2. upper left= place the legend in the upper left corner
# 3. lower left= place the legend in the lower left corner
# 4. lower right= place the legend in the lower right corner
# 5. right
# 6. center left
# 7. center right
# 8. lower center
# 9. upper center
# 10. center

plt.grid(True, color ='gray', linestyle = '--', linewidth= 0.5, alpha=0.7)
plt.show()
```

## **HeatMap**

```python
# 1. Positive Coorelation - When one variable increases the other variable also tends to increase - Height and weight often tends to have positive Correlation
# 2. Negative Correlation - When one variable increases the other variable tends to decrease.
# 3. Zero Correlation - No relationship between the variables, changes in one do not predict changes in the other

# Coorelations coefficients

# Pearson Correlation Coefficient (r) -
# 1 Continous variable ko linear relationship measure garna help garcha
# -1 Indicates a perfect negative correlation
# 0 Indicates no correlation

# Data Analysis
# Forecasting
# Risk Management

```

```python
plt.style.use('fivethirtyeight')

plt.figure(figsize=(10,6))
correlation = df.corr()
sns.heatmap(correlation, annot = True, cmap='coolwarm',linewidth=0.5)
plt.title('Correlation heatmap')
plt.show()
```

```python

plt.figure(figsize=(8,5))
plt.scatter(df['Year'], df['USA'],
                color = 'blue',
                alpha = 0.7,
                edgecolor ='black',
                s=120, # Size of the markers
                marker = 'o', # Shape of the marker
                linewidth = 1.5) # Adding linewidth for the edgecolor

plt.grid(color='grey',linestyle = '--', linewidth=0.8)
plt.gca().set_facecolor('lightblue')

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.xlabel("Year", fontsize=16)
plt.ylabel("Rate(USD)", fontsize = 16)
plt.title("Petrol Rates in the USA Over the years", fontsize=20, fontweight='bold', color='darkblue')

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
```

### **Bubble Chart**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

sizes_usa = df['USA'] * 500
sizes_canada = df['Canada'] * 500

plt.scatter(df['Year'], df['USA'],
            color='red',
            alpha=0.5,
            edgecolors='black',
            label='USA',
            s=sizes_usa,  # Size of the markers
            marker='o',  # Shape of the marker
            linewidth=1.5)  # Adding linewidth

plt.scatter(df['Year'], df['Canada'],
            color='green',
            alpha=0.5,
            edgecolors='black',
            label='Canada',
            s=sizes_canada,  # Corrected to sizes_canada
            marker='o',  # Shape of the marker
            linewidth=1.5)  # Adding linewidth

plt.xlabel("Year", fontsize=16)
plt.ylabel("Rate (USD)", fontsize=16)
plt.title("Petrol Rates in the USA and Canada Over the Years", fontsize=20, fontweight='bold', color='darkblue')
plt.legend()

plt.show()
```
